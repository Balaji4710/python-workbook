```python
with open("sample.txt", "w") as file:
    file.write("Hello, World!\n")
    file.write("Welcome to Python File Handling.")

print("Data written successfully.")
```

```python
with open("sample.txt", "r") as file:
    content = file.read()

print(content)
```

```python
with open("sample.txt", "a") as file:
    file.write("\nThis line was appended later.")

print("Data appended successfully.")
```

```python
with open("sample.txt", "r") as file:
    content = file.read()

characters = len(content)
words = len(content.split())
lines = len(content.splitlines())

print("Characters:", characters)
print("Words:", words)
print("Lines:", lines)
```

```python
with open("sample.txt", "r") as file:
    for line_number, line in enumerate(file, start=1):
        print(f"Line {line_number}: {line.strip()}")
```

```python
with open("sample.txt", "r") as source:
    content = source.read()

with open("destination.txt", "w") as destination:
    destination.write(content)

print("File copied successfully.")
```

```python
with open("sample.txt","r") as file:
    content=file.read()
    content=content.replace("Python","Java")
with open ("sample.txt","w") as file:
    file.write(content)
print ("Text Replaced Sucessfully")
```

```python
from collections import Counter


with open("sample.txt", "r") as file:
    words = file.read().split()

word_count = Counter(words)

for word, count in word_count.items():
    print(word, ":", count)
```

```python
with open("sample.txt", "r") as file:
    lines = file.readlines()

with open("reversed.txt", "w") as file:
    for line in reversed(lines):
        file.write(line)

print("File reversed successfully.")
```

```python
with open("sample.txt", "r") as f1:
    content1 = f1.read()

with open("destination.txt", "r") as f2:
    content2 = f2.read()

with open("merged.txt", "w") as merged:
    merged.write(content1)
    merged.write("\n")
    merged.write(content2)

print("Files merged successfully.")
```
